# First Fully Connected Layers with `MNIST` Data

```python
from tensorflow.keras import models 
from tensorflow.keras import layers
network = models.Sequential()
network.add(layers.Dense(512,activation='relu',name='Layer1',input_shape=(28*28,)))
network.add(layers.Dense(512,activation='relu',name='Layer2'))
network.add(layers.Dense(256,activation='relu',name='layer3'))
network.add(layers.Dense(50,activation='relu',name='Layer4'))
network.add(layers.Dense(10,activation='softmax',name='Layer5'))
network.summary()
```

# Second ConvNets with `MNIST` Data

```python
from tensorflow.keras import models, layers
model=models.Sequential()
model.add(layers.Conv2D(32,kernel_size=(3,3),activation='relu',name='Conv1',input_shape=(28,28,3)))
model.add(layers.MaxPooling2D(pool_size=(2,2),name='MaxPool1'))
model.add(layers.Conv2D(64,kernel_size=(3,3),activation='relu',name='Conv2'))
model.add(layers.MaxPooling2D(pool_size=(2,2),name='MaxPool2'))
model.add(layers.Conv2D(64,kernel_size=(3,3),activation='relu',name='Conv3'))
model.add(layers.Flatten(name='Flatten'))
model.add(layers.Dense(64,activation='relu',name='Dense1'))
model.add(layers.Dense(10,activation='softmax',name='Dense2'))
model.summary()
```

# Compare between No of features when using MaxPooling & Not using MaxPooling

```python
model_with_maxpool=models.Sequential()
model_with_maxpool.add(layers.Conv2D(32,kernel_size=(3,3),activation='relu',name='Conv1',input_shape=(28,28,3)))
model_with_maxpool.add(layers.MaxPooling2D(pool_size=(2,2),name='MaxPool1'))
model_with_maxpool.add(layers.Conv2D(64,kernel_size=(3,3),activation='relu',name='Conv2'))
model_with_maxpool.add(layers.MaxPooling2D(pool_size=(2,2),name='MaxPool2'))
model_with_maxpool.add(layers.Conv2D(64,kernel_size=(3,3),activation='relu',name='Conv3'))
model_with_maxpool.summary()
```

```python
model_no_maxpool=models.Sequential()
model_no_maxpool.add(layers.Conv2D(32,kernel_size=(3,3),activation='relu',name='Conv1',input_shape=(28,28,3)))
model_no_maxpool.add(layers.Conv2D(64,kernel_size=(3,3),activation='relu',name='Conv2'))
model_no_maxpool.add(layers.Conv2D(64,kernel_size=(3,3),activation='relu',name='Conv3'))
model_no_maxpool.summary()
```
